```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    "/root/autodl-tmp/fourlang_translation"
)

JUDGE_FILE = (
    PROJECT_ROOT
    / "results"
    / "teacher"
    / "madlad400_3b"
    / "qwen_judge"
    / "judge_details.csv"
)

df = pd.read_csv(JUDGE_FILE)

print(df.shape)
print(df.columns.tolist())
```

```python
en_uz_problem_df = df[
    (df["direction"] == "en_uz")
    &
    (
        df["grade"].isin(
            ["C", "D", "F", "PARSE_ERROR"]
        )
    )
].copy()

print("异常样本数：", len(en_uz_problem_df))
```
