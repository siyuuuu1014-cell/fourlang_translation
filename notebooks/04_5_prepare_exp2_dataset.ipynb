```python
opus_df = pd.read_csv(
    RAW_DIR
    /
    "opus"
    /
    "opus100_en_uz.csv"
)


print(
    opus_df.shape
)


opus_df.head()
```

```python
all_df = pd.concat(
    [
        hplt_df,
        tatoeba_df,
        opus_df
    ],
    ignore_index=True
)


print(
    all_df.shape
)
```

```python
def clean_text(text):

    if pd.isna(text):
        return ""

    text=str(text)

    text=text.strip()

    text=re.sub(
        r"\s+",
        " ",
        text
    )

    return text
```

```python
all_df["src_en"] = (
    all_df["src_en"]
    .apply(clean_text)
)


all_df["tgt_uz"] = (
    all_df["tgt_uz"]
    .apply(clean_text)
)
```

```python
all_df = all_df[
    (all_df.src_en!="")
    &
    (all_df.tgt_uz!="")
]


print(
    all_df.shape
)
```

```python
def length_filter(row):

    en_len=len(
        row.src_en.split()
    )

    uz_len=len(
        row.tgt_uz.split()
    )

    return (
        2 <= en_len <= 50
        and
        2 <= uz_len <= 50
    )


all_df = all_df[
    all_df.apply(
        length_filter,
        axis=1
    )
]


print(
    all_df.shape
)
```

```python
all_df = all_df.drop_duplicates(
    subset=[
        "src_en",
        "tgt_uz"
    ]
)


print(
    all_df.shape
)
```

```python
all_df = all_df.drop_duplicates(
    subset=[
        "src_en"
    ]
)


print(
    all_df.shape
)
```

```python
all_df.sample(
    20,
    random_state=42
)
```

```python
parallel_df = all_df.sample(
    n=50000,
    random_state=42
)


print(
    parallel_df.shape
)
```

```python
en_uz = pd.DataFrame({

    "src_lang":"en",

    "tgt_lang":"uz",

    "src_text":
        parallel_df["src_en"],

    "tgt_text":
        parallel_df["tgt_uz"]

})
```

```python
uz_en = pd.DataFrame({

    "src_lang":"uz",

    "tgt_lang":"en",

    "src_text":
        parallel_df["tgt_uz"],

    "tgt_text":
        parallel_df["src_en"]

})
```

```python
directional_df = pd.concat(
    [
        en_uz,
        uz_en
    ],
    ignore_index=True
)


print(
    directional_df.shape
)
```

```python
directional_df = directional_df.sample(
    frac=1,
    random_state=42
)


train_size=int(
    len(directional_df)*0.9
)


valid_size=int(
    len(directional_df)*0.05
)


train_df = directional_df.iloc[
    :train_size
]


valid_df = directional_df.iloc[
    train_size:
    train_size+valid_size
]


test_df = directional_df.iloc[
    train_size+valid_size:
]


print(
    train_df.shape,
    valid_df.shape,
    test_df.shape
)
```

```python
train_df.to_json(
    EXP2_DIR/
    "train_50k.jsonl",

    orient="records",

    lines=True,

    force_ascii=False
)


valid_df.to_json(
    EXP2_DIR/
    "validation.jsonl",

    orient="records",

    lines=True,

    force_ascii=False
)


test_df.to_json(
    EXP2_DIR/
    "test.jsonl",

    orient="records",

    lines=True,

    force_ascii=False
)
```
