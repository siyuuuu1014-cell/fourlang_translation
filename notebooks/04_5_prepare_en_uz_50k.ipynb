```python
from pathlib import Path
import pandas as pd


PROJECT_ROOT = Path(
    r"D:\dev\projects\fourlang_translation"
)


SAVE_DIR = (
    PROJECT_ROOT
    /
    "data"
    /
    "raw"
    /
    "en_uz"
    /
    "opus"
)


SAVE_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
train = dataset["train"]


rows=[]


for item in train:

    rows.append(
        {
            "src_en":
                item["translation"]["en"],

            "tgt_uz":
                item["translation"]["uz"]
        }
    )


opus_df = pd.DataFrame(rows)


print(opus_df.shape)

opus_df.head()
```

```python
opus_df.to_csv(
    SAVE_DIR /
    "opus100_en_uz.csv",
    index=False,
    encoding="utf-8-sig"
)
```
