```python
from pathlib import Path
import pandas as pd

DOWNLOAD_DIR = Path(r"D:\Download")

# 自动找你刚下载的 en-uz tsv，避免手敲中文文件名出错
tsv_files = list(DOWNLOAD_DIR.glob("*英语*乌兹别克语*.tsv"))

print("找到的文件：")
for f in tsv_files:
    print(f)

assert len(tsv_files) > 0, "没有找到 TSV 文件"

tsv_path = tsv_files[0]

print("\n当前使用：")
print(tsv_path)
```

```python
raw_df = pd.read_csv(
    tsv_path,
    sep="\t",
    header=None,
    dtype=str,
    keep_default_na=False,
)

print("数据形状：", raw_df.shape)

raw_df.head(20)
```

```python
print("列数：", raw_df.shape[1])

for i in range(raw_df.shape[1]):
    print(f"\n===== 第 {i} 列 =====")
    print(raw_df.iloc[:10, i].tolist())
```

```python
df = raw_df.rename(
    columns={
        0: "en_id",
        1: "en",
        2: "uz_id",
        3: "uz",
    }
).copy()

df.head(20)
```

```python
print(df.columns.tolist())
print(df.shape)
```

```python
print("总数据量:", len(df))

print(
    "English 空文本:",
    (df["en"].str.strip() == "").sum()
)

print(
    "Uzbek 空文本:",
    (df["uz"].str.strip() == "").sum()
)
```

```python
print(df.isna().sum())
```

```python
pair_duplicates = df.duplicated(
    subset=["en", "uz"]
).sum()

print("完全重复 pair:", pair_duplicates)
```

```python
print(
    "重复 English:",
    df.duplicated(subset=["en"]).sum()
)

print(
    "重复 Uzbek:",
    df.duplicated(subset=["uz"]).sum()
)
```

```python
df = df.drop_duplicates(
    subset=["en", "uz"]
).reset_index(drop=True)

print("去重后:", len(df))
```

```python
import re

CYRILLIC_PATTERN = re.compile(
    r"[\u0400-\u04FF]"
)

def has_cyrillic(text):
    return bool(
        CYRILLIC_PATTERN.search(str(text))
    )

df["uz_has_cyrillic"] = (
    df["uz"].apply(has_cyrillic)
)

df["uz_has_cyrillic"].value_counts()
```

```python
latin_count = (~df["uz_has_cyrillic"]).sum()
cyrillic_count = df["uz_has_cyrillic"].sum()

print("Latin Uzbek:", latin_count)
print("Cyrillic Uzbek:", cyrillic_count)

print(
    "Latin比例:",
    round(latin_count / len(df) * 100, 2),
    "%"
)

print(
    "Cyrillic比例:",
    round(cyrillic_count / len(df) * 100, 2),
    "%"
)
```

```python
cyrillic_df = df[
    df["uz_has_cyrillic"]
].copy()

cyrillic_df[
    ["en", "uz"]
].head(30)
```

```python
latin_df = df[
    ~df["uz_has_cyrillic"]
].copy()

latin_df[
    ["en", "uz"]
].head(30)
```
